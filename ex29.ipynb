```python
import bisect

def max_profit_intervals(x, r):
    n = len(x)
    points = sorted(zip(x, r))
    xs = [0] + [p[0] for p in points]   
    rs = [0] + [p[1] for p in points]

    compat = [0] * (n + 1)
    for i in range(1, n + 1):
        j = bisect.bisect_left(xs, xs[i] - 5, 1, i)
        compat[i] = j - 1   

    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        take = rs[i] + dp[compat[i]]   
        skip = dp[i - 1]               
        dp[i] = max(skip, take)

    selected = []
    i = n
    while i > 0:
        if dp[i] == dp[i - 1]:
            i -= 1          
        else:
            selected.append(xs[i])
            i = compat[i]   

    selected.reverse()
    return dp[n], selected


x = [6, 7, 12, 14]
r = [5, 6, 5, 1]
profit, chosen = max_profit_intervals(x, r)
print("Максимальная прибыль:", profit)
print("Координаты выбранных щитов:", chosen)
```
