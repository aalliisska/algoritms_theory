```python
def max_profit_intervals(x, r):
    intervals = [(x[i], x[i] + 5, r[i]) for i in range(len(x))]
    intervals.sort(key=lambda iv: iv[1])  
    
    n = len(intervals)
    dp = [0] * (n + 1)
    
    for i in range(1, n + 1):
        start, end, profit = intervals[i-1]
        
        prev = -1
        for j in range(i-2, -1, -1):
            if intervals[j][1] < start: 
                prev = j
                break
        
        if prev == -1:
            take = profit
        else:
            take = profit + dp[prev + 1]
        
        dp[i] = max(dp[i-1], take)
    
    return dp[n]
```

```python
x = [6, 7, 12, 14]
r = [5, 6, 5, 1]
max_profit_intervals(x, r)
```
