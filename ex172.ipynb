```python
def optimal_sell_order(rates):
    arr = []
    for i in range(len(rates)):
        arr.append((i, rates[i]))
    
    arr.sort(key=lambda x: x[1])
    
    res = []
    for a in arr:
        res.append(a[0])
    return res


def calc_revenue(rates, order):
    total = 0
    for t in range(len(order)):
        idx = order[t]
        total += 100 * (rates[idx] ** t)
    return total



rates = [0.9, 0.5, 0.8, 0.3]

order = optimal_sell_order(rates)
order
    
money = calc_revenue(rates, order)
money
    
bad = order[::-1]
bad_money = calc_revenue(rates, bad)
bad_money
    
for t in range(len(order)):
    i = order[t]
    price = 100 * (rates[i] ** t)
    print(f"  месяц {t}: оборудование {i}, r={rates[i]}, цена={price:.2f}")
```
