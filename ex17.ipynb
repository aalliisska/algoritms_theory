```python
def optimal_purchase_order(rates):
    items = []
    for i in range(len(rates)):
        items.append((i, rates[i]))
    
    items.sort(key=lambda x: x[1], reverse=True)

    result = []
    for item in items:
        result.append(item[0])
    
    return result


def calculate_total_cost(rates, order):
    total = 0
    for month in range(len(order)):
        license_idx = order[month]
        price = 100 * (rates[license_idx] ** month)
        total += price
    return total



rates = [1.2, 1.5, 1.8, 1.3]
best = optimal_purchase_order(rates)
print("Оптимальный порядок покупки:", best)
    
cost = calculate_total_cost(rates, best)
print(f"Общая стоимость: {cost:.2f}")
    
    
worst = best[::-1] 
worst_cost = calculate_total_cost(rates, worst)
print(f"Если наоборот: {worst_cost:.2f}")
    

for month in range(len(best)):
    idx = best[month]
    price = 100 * (rates[idx] ** month)
    print(f"  Месяц {month}: лицензия {idx} (r={rates[idx]}), цена={price:.2f}")
```
