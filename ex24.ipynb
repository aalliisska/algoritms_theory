```python
from collections import deque

def kruskal(n, edges):
    edges = sorted(edges, key=lambda e: e[2])
    parent = list(range(n))
    rank = [0] * n

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    def union(x, y):
        x, y = find(x), find(y)
        if x == y:
            return False
        if rank[x] < rank[y]:
            parent[x] = y
        elif rank[x] > rank[y]:
            parent[y] = x
        else:
            parent[y] = x
            rank[x] += 1
        return True

    mst = []
    cost = 0
    for u, v, w in edges:
        if union(u, v):
            mst.append((u, v, w))
            cost += w
            if len(mst) == n - 1:
                break
    return mst, cost


def build_tree(mst, n):
    tree = [[] for _ in range(n)]
    for u, v, w in mst:
        tree[u].append((v, w))
        tree[v].append((u, w))
    return tree


def find_max_on_path(tree, n, a, b):
    parent = [-1] * n
    max_edge = [(-1, -1, 0)] * n  
    q = deque([a])
    parent[a] = a

    while q:
        v = q.popleft()
        if v == b:
            break
        for to, w in tree[v]:
            if parent[to] == -1:
                parent[to] = v
                if w > max_edge[v][2]:
                    max_edge[to] = (v, to, w)
                else:
                    max_edge[to] = max_edge[v]
                q.append(to)

    return max_edge[b]  


def check_mst(tree, n, v, w, new_cost):
    max_w, _, _ = find_max_on_path(tree, n, v, w)
    return new_cost >= max_w





n = 4
edges = [
    (0, 1, 1),
    (1, 2, 2),
    (2, 3, 3),
    (0, 3, 4),
]


mst, cost = kruskal(n, edges)
mst, cost

tree = build_tree(mst, n)

check_mst(tree, n, 0, 3, 5)
```
