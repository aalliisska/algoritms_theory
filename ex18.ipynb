```python
def is_subsequence(s, sub):
    j = 0
    for x in s:
        if j < len(sub) and x == sub[j]:
            j += 1
        if j == len(sub):
            return True

    return j == len(sub)


tests = [
    (["купить Amazon", "купить Yahoo", "купить eBay", "купить Yahoo", "купить Yahoo", "купить Oracle"],
    ["купить Yahoo", "купить eBay", "купить Yahoo", "купить Oracle"], True),
    
    (["купить AAPL", "купить GOOG", "продать AAPL", "купить TSLA", "купить GOOG"],
    ["купить AAPL", "купить GOOG"], True),  # AAPL потом GOOG
]

for i, (s, sub, expected) in enumerate(tests, 1):
    res = is_subsequence(s, sub)
    ok = "ok" if res == expected else "fail"
    s_str = str(s[:3]) + "..." + str(s[-2:]) if len(s) > 5 else str(s)
    sub_str = str(sub[:3]) + "..." + str(sub[-2:]) if len(sub) > 5 else str(sub)
    print(f"{ok} тест {i}: {sub_str} в {s_str} -> {res} (должно быть {expected})")
```
