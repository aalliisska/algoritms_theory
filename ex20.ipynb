```python
def triathlon_order(swim, bike, run):
    n = len(swim)
    
    q = []
    for i in range(n):
        q.append(bike[i] + run[i])
    
    first = []
    second = []
    for i in range(n):
        if swim[i] <= q[i]:
            first.append((i, swim[i], q[i]))
        else:
            second.append((i, swim[i], q[i]))
    
    first.sort(key=lambda x: x[1])
    second.sort(key=lambda x: -x[2])
    
    order = []
    for x in first:
        order.append(x[0])
    for x in second:
        order.append(x[0])
    
    return order


swim = [5, 6, 3, 4]
bike = [4, 2, 5, 8]
run = [3, 1, 2, 1]

order = triathlon_order(swim, bike, run)
order
```
