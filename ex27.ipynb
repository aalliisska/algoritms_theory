```python
def find_peak(A):
    n = len(A)
    left, right = 0, n - 1
    
    while left < right:
        mid = (left + right) // 2
        if A[mid] < A[mid + 1]:
            left = mid + 1
        else:
            right = mid
    
    return left


arr1 = [1, 3, 5, 7, 9, 8, 6, 4, 2]
peak1 = find_peak(arr1)
peak1,  arr1[peak1]
    
```
