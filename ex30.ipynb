```python
def max_weight(weights):
    n = len(weights)
    if n == 0:
        return 0
    if n == 1:
        return weights[0]
    
    dp = [0] * (n + 1)
    dp[1] = weights[0]
    
    for i in range(2, n + 1):
        take = weights[i-1] + dp[i-2]
        skip = dp[i-1]
        dp[i] = max(take, skip)
    
    return dp[n]

weights = [5, 6, 5, 1]
max_weight(weights)
```

```python
def max_weight_independent_set(weights):
    n = len(weights)
    if n == 0:
        return 0, []
    if n == 1:
        return weights[0], [0]

    dp = [0] * n
    dp[0] = weights[0]
    dp[1] = max(weights[0], weights[1])

    for i in range(2, n):
        dp[i] = max(dp[i-1], dp[i-2] + weights[i])

    selected = []
    i = n - 1
    while i >= 0:
        if i == 0:
            selected.append(0)
            break
        if i == 1:
            if dp[1] == weights[1] and (len(selected) == 0 or selected[-1] != 0):
                selected.append(1)
            else:
                selected.append(0)
            break
        if dp[i] == dp[i-1]:
            i -= 1
        else:
            selected.append(i)
            i -= 2
    selected.reverse()
    return dp[n-1], selected



weights = [2, 3, 2]
max_weight, indices = max_weight_independent_set(weights)
max_weight
```

Контр пример, что алгоритм «приоритетного выбора максимального веса» не всегда оптимален.

Рассмотрим путевой граф:

вершина 0: вес 2
вершина 1: вес 3
вершина 2: вес 2

Алгоритм:

Выбирает вершину с максимальным весом — вершину 1 (вес 3).
Добавляет её в множество S.
Удаляет вершину 1 и её соседей (вершины 0 и 2).
Граф становится пустым, алгоритм завершается.
Результат: S = {1}, общий вес = 3.

Однако оптимальное независимое множество — {0, 2} с общим весом 2 + 2 = 4.

Таким образом, жадный алгоритм даёт вес 3 вместо максимально возможного 4.

контр пример: алгоритм выбирает вершину с максимальным весом,
сначала берёт вершину 1 (вес 5), удаляет её и соседей 0, 2, остаётся только вершина 3 (вес 1), берёт её. итог: вершины [1, 3], вес 5 + 1 = 6. Но оптимально — взять вершины 0 и 2: вес 4 + 4 = 8. алгоритм проигрывает: 6 < 8, потому что взял 5, но потерял дважды по 4

```python
def max_of_parities(weights):
    s1 = 0  
    for i in range(0, len(weights), 2): 
        s1 += weights[i]
    
    s2 = 0
    for i in range(1, len(weights), 2):  
        s2 += weights[i]

    if s1 > s2:
        return s1
    else:
        return s2


def optimal(weights):
    n = len(weights)
    if n == 0:
        return 0
    if n == 1:
        return weights[0]
    
    dp = [0] * n
    dp[0] = weights[0]
    dp[1] = max(weights[0], weights[1])
    
    for i in range(2, n):
        dp[i] = max(dp[i-1], dp[i-2] + weights[i])
    
    return dp[n-1]  


weights = [1, 100, 1, 1, 100]

max_of_parities(weights)
```

```python
def max_weight_independent_set_path(weights):
    n = len(weights)
    if n == 0:
        return 0, []
    if n == 1:
        return weights[0], [1]

    dp = [0] * n
    dp[0] = weights[0]
    dp[1] = max(weights[0], weights[1])

    for i in range(2, n):
        dp[i] = max(dp[i-1], dp[i-2] + weights[i])

    selected = []
    i = n - 1
    while i >= 0:
        if i == 0:
            selected.append(0)
            break
        if i == 1:
            if dp[1] == weights[1] and weights[1] >= weights[0]:
                selected.append(1)
            else:
                selected.append(0)
            break
        if dp[i] == dp[i-1]:
            i -= 1
        else:
            selected.append(i)
            i -= 2

    result = [idx + 1 for idx in selected]
    result.sort()
    
    return dp[-1], result


w1 = [2, 3, 2]
w, nodes = max_weight_independent_set_path(w1)
print(f"веса {w1} -> макс вес {w}, узлы {nodes}")
```
