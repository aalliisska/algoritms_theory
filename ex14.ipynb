```python
def check(n, edges):
    parent = list(range(n))
    xor = [0] * n      
    rank = [0] * n
    
    def find(x):
        if parent[x] != x:
            root = find(parent[x])
            xor[x] ^= xor[parent[x]]
            parent[x] = root
        return parent[x]
    
    for u, v, w in edges:
        ru = find(u)
        rv = find(v)
        
        if ru == rv:
            if (xor[u] ^ xor[v]) != w:
                return False
            continue
        
        
        if rank[ru] < rank[rv]:
            ru, rv = rv, ru
            u, v = v, u  
        
        parent[rv] = ru
        xor[rv] = xor[u] ^ xor[v] ^ w
        if rank[ru] == rank[rv]:
            rank[ru] += 1
    
    return True


print(check(4, [(0,1,0), (1,2,1), (2,3,0)])) 
print(check(3, [(0,1,0), (1,2,0), (0,2,1)]))  
print(check(3, [(0,1,1), (1,2,1), (2,0,1)]))   
print(check(5, [(0,1,1), (1,2,1), (2,3,1), (3,4,1), (0,4,0)])) 
```
